## How Colab works

A Colab notebook is a sequence of cells. Text cells, like this one, explain. Code cells hold Python instructions: Select a code cell and press Shift+Enter to run it, and its output appears directly below it. Save your own copy first with File, Save a copy in Drive. The cells run in order from top to bottom, and a value created in one cell can be used in every later cell. Nothing can break: If something goes wrong, choose Runtime, Restart session, and run the cells again from the top. Run the next cell now.

```python
# Python practice: a small example for learning the language
print("Hello from the clinic app!")
print(2 + 3, "patients are waiting")
```

The next cell defines a small helper, `check`, which compares an answer with the expected value and prints immediate feedback. The cell after it keeps the expected answers, with its code hidden.

```python
def check(name, value, expected=None, tol=1e-6):
    """Compares an answer with the expected value and prints immediate feedback.
    Without an expected value in the call, check takes the value and its tolerance
    from ANSWERS, the table of the next cell."""
    if expected is None:
        answer, tol = ANSWERS[name]
        expected = answer()
    if value is None:
        print(f"{name}: not answered yet. Replace None and run the cell again.")
        return False
    try:
        scale = max(1.0, abs(float(expected)))
        ok = abs(float(value) - float(expected)) <= tol * scale
    except (TypeError, ValueError):
        ok = value == expected
    print(f"{name}: " + ("correct." if ok else
                         "not yet. Compare your steps with the text above."))
    return ok
```

```python
#@title Expected answers of the exercises (run this cell; open it only after trying the exercises)
ANSWERS = {
    "Exercise 0.1": (lambda: 105, 1e-6),
    "Exercise 0.2": (lambda: 3, 1e-6),
    "Exercise 0.3": (lambda: 0.729, 1e-6),
    "Exercise 0.4": (lambda: 46.5, 1e-6),
}
```

## Variables and types

A variable is a name for a value, created with `=`. Whole numbers are integers, numbers with a decimal point are floats, text in quotation marks is a string, and `True` or `False` is a boolean.

```python
# Python practice: a small example for learning the language
slots_per_hour = 4        # an integer
minutes_per_slot = 15.0   # a float
clinic = "Cardiology"     # a string
open_today = True         # a boolean
print(clinic, "offers", slots_per_hour * 8, "slots in eight hours")
print(type(slots_per_hour), type(minutes_per_slot))
print(type(clinic), type(open_today))
```

**Exercise 0.1.** A clinic offers 15 slots a day on 7 days. Store the number of slots in a week in `slots_week`.

```python
slots_week = None   # your answer

check("Exercise 0.1", slots_week)
```

```python
#@title Reference solution (run this cell; open it only after trying)
slots_week = 15 * 7

check("Exercise 0.1", slots_week)
```

## Lists, loops and decisions

A list keeps values in order between square brackets. A `for` loop takes one value after another, and an `if` statement runs its indented block only when its condition is true.

```python
# Python practice: a small example for learning the language
waits_min = [12, 35, 8, 41, 27, 33]
long = 0
for w in waits_min:
    if w > 30:
        long = long + 1
        print(w, "minutes: a long wait")
print("long waits:", long, "of", len(waits_min))
```

**Exercise 0.2.** Count the waits in `waits_min` that are shorter than 30 minutes and store the count in `n_short`.

```python
n_short = None   # your answer

check("Exercise 0.2", n_short)
```

```python
#@title Reference solution (run this cell; open it only after trying)
n_short = 0
for w in waits_min:
    if w < 30:
        n_short = n_short + 1

check("Exercise 0.2", n_short)
```

## Functions

A function is a named piece of code with parameters. `def` defines it, and `return` hands back its result, which can then be used like any other value.

```python
# Python practice: a small example for learning the language
def p_no_slip(e, steps):
    """Probability of no slip in a number of steps, each with slip probability e."""
    return (1 - e) ** steps

print(round(p_no_slip(0.05, 10), 3))
```

**Exercise 0.3.** Use `p_no_slip` to store the probability of no slip in three steps with $e = 0.1$ in `p3`. The operator `**` raises to a power.

```python
p3 = None   # your answer

check("Exercise 0.3", p3)
```

```python
#@title Reference solution (run this cell; open it only after trying)
p3 = p_no_slip(0.1, 3)

check("Exercise 0.3", p3)
```

## Three libraries: NumPy, pandas and Matplotlib

NumPy computes with whole arrays of numbers at once, pandas keeps tables with named columns, and Matplotlib draws figures. `import ... as ...` loads a library under a short name.

```python
# Python practice: a small example for learning the language
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ages = np.array([74, 35, 52, 41, 68, 29])
print("mean age:", ages.mean())
table = pd.DataFrame({"patient": ["A", "B", "C", "D", "E", "F"], "age": ages,
                      "minutes": [6.1, 2.4, 3.9, 3.0, 5.2, 2.2]})
print(table)
plt.scatter(table["age"], table["minutes"])
plt.xlabel("age (years)")
plt.ylabel("booking time (min)")
plt.title("Older patients take longer in this invented example")
plt.show()
```

**Exercise 0.4.** Use NumPy to store the median of `ages` in `age_median`. The function is `np.median`.

```python
age_median = None   # your answer

check("Exercise 0.4", age_median)
```

```python
#@title Reference solution (run this cell; open it only after trying)
age_median = np.median(ages)

check("Exercise 0.4", age_median)
```
